## Problem 01: Reverse String
> **Problem Statement**: Given a string `s`, return the reversed string without using Python's built-in `s[::-1]` slicing.

### Approach:
* Strings in Python are immutable, so convert `s` into a list of characters.
* Use two pointers (`left = 0`, `right = len(s) - 1`).
* Swap characters while `left < right` and increment/decrement pointers.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(n)$ (to store the mutable character list)


```python
def reverse_string(s: str) -> str:
    chars = list(s)
    left, right = 0, len(chars) - 1
    
    while left < right:
        chars[left], chars[right] = chars[right], chars[left]
        left += 1
        right -= 1
        
    return "".join(chars)

# Test cases:
test_cases = ["hello", "Python", "racecar", ""]
for s in test_cases:
    print(f"reverse_string({repr(s)}) -> {repr(reverse_string(s))}")
    assert reverse_string(s) == s[::-1], f"Mismatch for {s}"

print("✅ Reverse string tests passed!")
```

## Problem 02: Valid Palindrome
> **Problem Statement**: Given a string `s`, determine if it is a palindrome, considering only alphanumeric characters and ignoring cases.

### Approach:
* Use two pointers at both ends.
* Skip non-alphanumeric characters using `.isalnum()`.
* Compare matching characters in lowercase.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(1)$


```python
def is_palindrome(s: str) -> bool:
    left, right = 0, len(s) - 1
    
    while left < right:
        # Skip non-alphanumeric from left
        while left < right and not s[left].isalnum():
            left += 1
        # Skip non-alphanumeric from right
        while left < right and not s[right].isalnum():
            right -= 1
            
        if s[left].lower() != s[right].lower():
            return False
            
        left += 1
        right -= 1
        
    return True

# Test cases:
tests = [
    ("A man, a plan, a canal: Panama", True),
    ("race a car", False),
    (" ", True),
    ("Was it a car or a cat I saw?", True)
]

for s, expected in tests:
    result = is_palindrome(s)
    print(f"is_palindrome('{s}') = {result} (Expected: {expected})")
    assert result == expected

print("✅ Palindrome tests passed!")
```

## Problem 09: First Non-Repeating Character
> **Problem Statement**: Given a string `s`, find the first non-repeating character in it and return its index. If it doesn't exist, return `-1`.

### Approach:
* Use `collections.Counter` or a dictionary to count character frequencies in $\mathcal{O}(n)$.
* Scan `s` a second time: the first character with frequency `1` is the answer.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(1)$ (bounded by 26 English lowercase letters)


```python
from collections import Counter

def first_uniq_char(s: str) -> int:
    counts = Counter(s)
    
    for i, char in enumerate(s):
        if counts[char] == 1:
            return i
            
    return -1

# Test cases:
tests = [
    ("leetcode", 0),      # 'l' is unique at index 0
    ("loveleetcode", 2),  # 'v' is unique at index 2
    ("aabb", -1)          # No unique character
]

for s, expected in tests:
    result = first_uniq_char(s)
    print(f"first_uniq_char('{s}') -> {result} (Char: {s[result] if result != -1 else 'None'})")
    assert result == expected

print("✅ First non-repeating character tests passed!")
```

## Problem 13: Valid Anagram
> **Problem Statement**: Given two strings `s` and `t`, return `True` if `t` is an anagram of `s`, and `False` otherwise.

### Approach:
* Lengths must match; otherwise return `False`.
* Count character frequencies using a hash map or single array.
* Compare frequency dictionaries.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(1)$ (at most 26 keys)


```python
def is_anagram(s: str, t: str) -> bool:
    if len(s) != len(t):
        return False
        
    counts = {}
    for char in s:
        counts[char] = counts.get(char, 0) + 1
        
    for char in t:
        if char not in counts or counts[char] == 0:
            return False
        counts[char] -= 1
        
    return True

# Test cases:
tests = [
    ("anagram", "nagaram", True),
    ("rat", "car", False),
    ("listen", "silent", True),
    ("a", "ab", False)
]

for s, t, expected in tests:
    result = is_anagram(s, t)
    print(f"is_anagram('{s}', '{t}') -> {result}")
    assert result == expected

print("✅ Anagram tests passed!")
```

## 💡 Practice Challenge: Group Anagrams
Try implementing a function to group anagrams together:
e.g. `["eat","tea","tan","ate","nat","bat"]` -> `[['eat', 'tea', 'ate'], ['tan', 'nat'], ['bat']]`


```python
from collections import defaultdict

def group_anagrams(strs: list[str]) -> list[list[str]]:
    groups = defaultdict(list)
    for word in strs:
        key = "".join(sorted(word))
        groups[key].append(word)
    return list(groups.values())

words = ["eat", "tea", "tan", "ate", "nat", "bat"]
grouped = group_anagrams(words)
print("Grouped Anagrams:", grouped)
```
