## 1. Defining `ListNode` & Helper Utilities
Let's build a clean `ListNode` class and visual representation helper.


```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
        
    def __repr__(self):
        return f"ListNode({self.val})"

def build_linked_list(values: list) -> ListNode | None:
    """Converts a Python list into a Linked List chain."""
    if not values:
        return None
    head = ListNode(values[0])
    curr = head
    for v in values[1:]:
        curr.next = ListNode(v)
        curr = curr.next
    return head

def print_list(head: ListNode | None, max_nodes: int = 15) -> str:
    """Pretty-prints a linked list (with cycle limit safeguard)."""
    vals = []
    curr = head
    count = 0
    while curr and count < max_nodes:
        vals.append(str(curr.val))
        curr = curr.next
        count += 1
    if curr:
        vals.append("... (cycle detected or long list)")
    else:
        vals.append("None")
    return " -> ".join(vals)

# Example:
my_list = build_linked_list([10, 20, 30, 40])
print("Built list:", print_list(my_list))
```

## Problem 19: Detect Cycle in Linked List (Floyd's Tortoise & Hare)
> **Problem Statement**: Given `head` of a linked list, determine if the list has a cycle in it.

### Floyd's Algorithm Intuition:
* `slow` advances 1 step at a time: $S \leftarrow S.next$
* `fast` advances 2 steps at a time: $F \leftarrow F.next.next$
* If there is a cycle, the faster pointer will eventually lap the slower pointer and they will point to the exact same node (`slow is fast`).
* If `fast` reaches `None`, the list is acyclic.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(1)$


```python
def has_cycle(head: ListNode | None) -> bool:
    if not head or not head.next:
        return False
        
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
        if slow == fast:
            return True # Cycle detected!
            
    return False

# Test 1: Acyclic list: 1 -> 2 -> 3 -> 4 -> None
acyclic = build_linked_list([1, 2, 3, 4])
print("Acyclic List:", print_list(acyclic))
print("Has cycle?", has_cycle(acyclic))
assert has_cycle(acyclic) is False

# Test 2: Cyclic list: 1 -> 2 -> 3 -> 4 -> points back to 2
n1, n2, n3, n4 = ListNode(1), ListNode(2), ListNode(3), ListNode(4)
n1.next, n2.next, n3.next, n4.next = n2, n3, n4, n2 # Cycle!

print("Has cycle in circular list?", has_cycle(n1))
assert has_cycle(n1) is True
print("✅ Cycle detection tests passed!")
```

## Problem 20: Middle of the Linked List
> **Problem Statement**: Given the `head` of a singly linked list, return the middle node. If there are two middle nodes, return the second middle node.

### Approach:
* `slow` moves 1 step; `fast` moves 2 steps.
* When `fast` reaches the end (`fast is None` or `fast.next is None`), `slow` is exactly at the middle node!

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(1)$


```python
def find_middle(head: ListNode | None) -> ListNode | None:
    slow = head
    fast = head
    
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        
    return slow

# Odd length: [1, 2, 3, 4, 5] -> Middle is 3
odd_list = build_linked_list([1, 2, 3, 4, 5])
mid_odd = find_middle(odd_list)
print(f"Middle of [1,2,3,4,5]: {mid_odd.val if mid_odd else None}")
assert mid_odd.val == 3

# Even length: [1, 2, 3, 4, 5, 6] -> Middle is 4 (second middle)
even_list = build_linked_list([1, 2, 3, 4, 5, 6])
mid_even = find_middle(even_list)
print(f"Middle of [1,2,3,4,5,6]: {mid_even.val if mid_even else None}")
assert mid_even.val == 4

print("✅ Find middle tests passed!")
```

## 💡 Bonus Challenge: Reverse a Singly Linked List
> **Problem Statement**: Reverse the arrows of a linked list in-place in $\mathcal{O}(n)$ time and $\mathcal{O}(1)$ space.


```python
def reverse_linked_list(head: ListNode | None) -> ListNode | None:
    prev = None
    curr = head
    
    while curr:
        nxt = curr.next # Save next
        curr.next = prev # Reverse link
        prev = curr     # Advance prev
        curr = nxt      # Advance curr
        
    return prev # New head

original = build_linked_list([1, 2, 3, 4, 5])
print("Original:", print_list(original))
reversed_head = reverse_linked_list(original)
print("Reversed:", print_list(reversed_head))
assert print_list(reversed_head) == "5 -> 4 -> 3 -> 2 -> 1 -> None"
print("🎉 All linked list tests passed successfully!")
```
