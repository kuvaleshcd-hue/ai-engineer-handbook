## Problem 08: Two Sum (Classic Must-Know)
> **Problem Statement**: Given an array of integers `nums` and an integer `target`, return indices of the two numbers such that they add up to `target`.

### Approach:
* As you iterate through `nums`, calculate the needed `complement = target - num`.
* Check if `complement` is already in your `seen` map (`val -> index`).
* If found, return `[seen[complement], current_index]`.

* **Time Complexity**: $\mathcal{O}(n)$
* **Space Complexity**: $\mathcal{O}(n)$


```python
def two_sum(nums: list[int], target: int) -> list[int]:
    seen = {} # value -> index
    
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
        
    return []

# Test cases:
tests = [
    ([2, 7, 11, 15], 9, [0, 1]),
    ([3, 2, 4], 6, [1, 2]),
    ([3, 3], 6, [0, 1])
]

for nums, target, expected in tests:
    result = two_sum(nums, target)
    print(f"two_sum({nums}, target={target}) -> {result}")
    assert result == expected

print("✅ Two Sum tests passed!")
```

## Problem 06 & 16: Duplicate Management
* **Find Duplicates**: Return all values appearing more than once.
* **Remove Duplicates Preserving Order**: Deduplicate while keeping the first occurrence order intact.


```python
def find_duplicates(nums: list) -> list:
    """Finds all duplicate elements in a list. Time: O(n), Space: O(n)."""
    seen = set()
    duplicates = set()
    for x in nums:
        if x in seen:
            duplicates.add(x)
        else:
            seen.add(x)
    return list(duplicates)

def remove_duplicates_preserve_order(items: list) -> list:
    """Removes duplicates while maintaining original insertion order."""
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

sample = [1, 2, 3, 2, 4, 1, 5, 3]
print("Original:", sample)
print("Duplicates:", find_duplicates(sample))
print("Deduplicated (order preserved):", remove_duplicates_preserve_order(sample))

assert remove_duplicates_preserve_order([4, 5, 4, 1, 2, 5]) == [4, 5, 1, 2]
print("✅ Deduplication tests passed!")
```

## Problem 10: Merge Two Sorted Lists
> **Problem Statement**: Merge two already sorted lists into a single sorted list in $\mathcal{O}(n + m)$ time without re-sorting.


```python
def merge_sorted_lists(l1: list[int], l2: list[int]) -> list[int]:
    i, j = 0, 0
    merged = []
    
    while i < len(l1) and j < len(l2):
        if l1[i] <= l2[j]:
            merged.append(l1[i])
            i += 1
        else:
            merged.append(l2[j])
            j += 1
            
    # Append any remaining elements
    merged.extend(l1[i:])
    merged.extend(l2[j:])
    return merged

l1 = [1, 3, 5, 8]
l2 = [2, 4, 6, 7, 9]
result = merge_sorted_lists(l1, l2)
print("Merged:", result)
assert result == [1, 2, 3, 4, 5, 6, 7, 8, 9]
print("✅ Merge sorted lists passed!")
```

## Problem 11: Flatten Deeply Nested List
> **Problem Statement**: Flatten an arbitrarily nested list of elements into a single 1D list.


```python
def flatten_list(nested: list) -> list:
    result = []
    for item in nested:
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result

nested = [1, [2, [3, 4], 5], 6, [7, [8, [9]]]]
flat = flatten_list(nested)
print(f"Flattened {nested} -> {flat}")
assert flat == [1, 2, 3, 4, 5, 6, 7, 8, 9]
print("✅ Flatten list passed!")
```

## Problem 15 & 17: Max/Min Linear Scan & Dictionary Sorting


```python
def find_max_min(arr: list[int]) -> tuple[int, int]:
    """Finds (min, max) in a single linear pass O(n)."""
    if not arr:
        raise ValueError("List cannot be empty")
    curr_min = curr_max = arr[0]
    for val in arr[1:]:
        if val < curr_min:
            curr_min = val
        elif val > curr_max:
            curr_max = val
    return curr_min, curr_max

def sort_dict_by_value(d: dict, reverse: bool = False) -> dict:
    """Sorts dictionary entries by their values."""
    return dict(sorted(d.items(), key=lambda item: item[1], reverse=reverse))

scores = {"Alice": 88, "Bob": 95, "Charlie": 78, "Diana": 92}
print("Scores sorted by value (descending):", sort_dict_by_value(scores, reverse=True))

min_val, max_val = find_max_min([10, -5, 23, 0, 99, 4])
print(f"Min: {min_val}, Max: {max_val}")
assert (min_val, max_val) == (-5, 99)
print("✅ Max/Min and Dict sort passed!")
```
